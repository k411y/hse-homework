### Задача 1
```
Переведите содержимое файла purchase_log.txt в словарь purchases вида:
 {'1840e0b9d4': 'Продукты', ...}

Пример работы программы при выводе первых двух элементов словаря purchases:
1840e0b9d4 ‘Продукты‘
4e4f90fcfb  ‘Электроника‘
```

````python
import json

purchases = {}

with open("../src/purchase_log.txt", "r", encoding="utf-8") as file:
    next(file)      # Пропускаем заголовок

    for line in file:
        item = json.loads(line)
        purchases[item["user_id"]] = item["category"]
````

````python
# Проверка
for i, (key, value) in enumerate(purchases.items()):
    if i == 10:
        break

    print(key, value)
````

### Задача 2
```
Для каждого user_id в файле visit_log.csv определите третий столбец с категорией покупки. Если покупка была, сам файл visit_log.csv изменять не надо.
Запишите в файл funnel.csv визиты из файла visit_log.csv, в которых были покупки с указанием категории.

Примеры работы программы при выводе первых трёх строк файла funnel.csv:
user_id,source,category
1840e0b9d4,other,Продукты
4e4f90fcfb,context,Электроника
```

````python
import csv


with open("../src/visit_log.csv", "r", encoding="utf-8") as visit_file, \
     open("../src/funnel.csv", "w", encoding="utf-8", newline="") as funnel_file:
        reader = csv.DictReader(visit_file)
        writer = csv.writer(funnel_file)

        writer.writerow(["user_id", "source", "category"])
        for row in reader:
            if row["user_id"] in purchases:
                writer.writerow([row["user_id"], row["source"], purchases[row["user_id"]]])
````

````python
with open("../src/funnel.csv", "r", encoding="utf-8", newline="") as funnel_file:
    for i, line in enumerate(funnel_file):
        if i == 10:
            break

        print(line.rstrip())
````
